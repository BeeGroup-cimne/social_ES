# Population and Dwellings Census 2021 <a name="dwellings2021"></a>

The [2021 Population and Dwellings Census](https://www.ine.es/censos2021/C2021_Indicadores.csv) publishes, for every census tract in Spain, a fixed set of indicators describing its population (sex, average age, age groups, foreign nationality and birth origin, current and attained education, labour force status and marital status) together with its dwellings (total, main and non-main, tenure regime) and its households (total and by household size).

INE publishes this table at census-tract level only; the district and municipality tables are aggregated by `social_ES`, summing the counts and averaging each share weighting by the population it is a share of. Shares are returned as percentages (0-100) instead of the proportions of the source file. The indicators of the smallest census tracts are suppressed by INE for statistical confidentiality: those rows keep their population and have missing values elsewhere.

```python
dwellings_and_population_df = INE.DwellingsAndPopulationCensus(wd=wd)
```

# Empty and Secondary Dwellings Census (2001, 2011, 2021) <a name="emptydwellings"></a>

This dataset puts the empty, secondary and non-main dwelling counts of the [2001](https://www.ine.es/censo/es/inicio.jsp), [2011](https://www.ine.es/censos2011_datos/cen11_datos_resultados.htm) and [2021](https://www.ine.es/dyngs/INEbase/operacion.htm?c=Estadistica_C&cid=1254736177108&menu=resultados&idp=1254735572981) censuses in a single table, per municipality, province and autonomous community.

The 2021 census dropped the classification the two earlier ones used, so **two classifications are returned side by side rather than merged**:

* `Dwellings ~ Dwelling type:*` — the field-census classification, where an agent visiting the building sorted each dwelling into main, secondary (occasional use) or empty. `Main` and `Non-main` exist in all three censuses; `Secondary` and `Empty` only in 2001 and 2011, plus an `Other non-main` residual in 2001.
* `Dwellings ~ Electricity use:*` — what INE publishes instead for 2021, derived from the yearly electricity consumption of each dwelling: `Empty` (no supply contract, or less than the equivalent of 15 days a year), `Very low consumption` (up to 250 kWh), `Sporadic use` (251-750 kWh, one to three months a year) and `Regular use`.

A third group, `Dwellings ~ Comparable use:*`, aligns the two into one series that can be read across the three censuses, filled from whichever classification each census published:

| Comparable use | 2001 / 2011 (field census) | 2021 (electricity) |
|---|---|---|
| `Main` | `Dwelling type:Main` | `Electricity use:Regular use` |
| `Secondary` | `Dwelling type:Secondary` (+ `Other non-main`, in 2001) | `Electricity use:Very low consumption` + `Sporadic use` |
| `Empty` | `Dwelling type:Empty` | `Electricity use:Empty` |

The three classes partition the total in every census, so their shares always add to 100. Note that 2021's `Very low consumption` (about a month of use) counts as *secondary* rather than empty, and that 2001's `Other non-main` residual counts as secondary too — which is what INE does in its own 2001-2011 comparison.

**This is a bridge, not an identity.** The 2021 figures come from electricity meters and the earlier ones from a census agent's judgement at the door, so any movement along these lines between 2011 and 2021 mixes a real change in use with the change of instrument — most visibly in tourist municipalities, where a second home occupied for a full summer consumes well over 750 kWh and is classified as `Main`. The figures each census actually published stay available under `Dwelling type:*` and `Electricity use:*`.


```python
empty_and_secondary_dwellings_df = INE.EmptyAndSecondaryDwellingsCensus(wd=wd)
```

These are the geographical levels available:

```python
pprint.pp(list(empty_and_secondary_dwellings_df.keys()))
```

An example of the results for the municipality of Barcelona, across the three censuses, is the following:

```python
barcelona = empty_and_secondary_dwellings_df["Municipality"][
    empty_and_secondary_dwellings_df["Municipality"]["Municipality code"] == "08019"]
barcelona[["Year", "Dwellings",
           "Percentage of dwellings ~ Comparable use:Main",
           "Percentage of dwellings ~ Comparable use:Secondary",
           "Percentage of dwellings ~ Comparable use:Empty"]]
```

INE only publishes the 2011 secondary/empty split for municipalities over 2,000 inhabitants, and the 2021 electricity classification for those over 1,000 (about 97% of the population); the rest are `NaN` in the municipal table.

The province and autonomous community tables are complete regardless, because those columns are read from INE's own regional tables instead of being added up from the municipalities — summing the municipal table recovers only 85.9% of the 2021 empty dwellings nationally, and as little as 48% in Castilla y León:

```python
empty_and_secondary_dwellings_df["Autonomous community"][
    empty_and_secondary_dwellings_df["Autonomous community"]["Year"] == 2021][[
        "Autonomous community name", "Dwellings",
        "Dwellings ~ Electricity use:Empty",
        "Percentage of dwellings ~ Electricity use:Empty",
        "Percentage of dwellings ~ Electricity use:Sporadic use"]]
```

```python
INE.MapVariable(empty_and_secondary_dwellings_df['Municipality'].query('Year==2001'), wd=wd)
INE.MapVariable(empty_and_secondary_dwellings_df['Municipality'].query('Year==2011'), wd=wd)
INE.MapVariable(empty_and_secondary_dwellings_df['Municipality'].query('Year==2021'), wd=wd)
```

```python
pprint.pp(list(empty_and_secondary_dwellings_df['Municipality'].columns), width=400)
```

# Time Use Survey (EET 2009-2010) <a name="timeuse"></a>

This [dataset](https://www.ine.es/ftp/microdatos/emptiem/datos_emptiem0910.zip) provides weekly hourly activity shares and daily holiday/vacation probabilities derived from the Spanish Time Use Survey (Encuesta de Empleo del Tiempo, EET 2009-2010). The microdata are anonymised to the autonomous-community level and linked to census tracts via the HouseholdIncomeDistributionAtlas (2015 edition). Each census tract is assigned to an autonomous community and household income group based on its average net household income. The dataset returns three dictionaries:  (census tract → autonomous community + income band),  (activity shares per autonomous community, income group, weekday, and hour), and  (free/holiday-day probability per autonomous community, income group, and date). The survey records ten activity groups that sum to 100% within each (community, group, weekday, hour).


```python
time_use = INE.TimeUseSurvey(wd=wd)
pprint.pp(list(time_use.keys()))
```
